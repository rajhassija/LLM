```python
# convert json to owl

from owlready2 import *

# Create a new ontology
ontology = get_ontology("http://example.com/ontology/")

with ontology:
    # Define classes
    class Product(Thing): pass
    class plan(Thing): pass
    class feature(Thing): pass
    class Order(Thing): pass
    class Customer(Thing): pass
    class requestransaction(Thing): pass
    class requestoutcome(Thing): pass

    # Define properties for request transaction 
    class has_reqid(requestransaction >> int): pass
    class has_reqamount(requestransaction >> int): pass
    class has_reqdisttype(requestransaction >> int): pass

    # Define properties for request outcome 
    class has_reqid(requestoutcome >> int): pass
    class has_addresschangedaysoutcome(requestoutcome >> int): pass
    class has_addresschangedaysoutcome(requestoutcome >> int): pass
    

    # Define properties for Product
    class has_id(Product >> str): pass
    class has_name(Product >> str): pass
    class has_price(Product >> float): pass
    class has_category(Product >> str): pass

    # Define properties for plan
    class has_id(plan >> str): pass
    class has_name(plan >> str): pass

    # Define properties for Order
    class has_order_id(Order >> str): pass
    class has_product_id(Order >> str): pass
    class has_quantity(Order >> int): pass
    class has_total_price(Order >> float): pass

    # Define properties for Customer
    class has_customer_id(Customer >> str): pass
    class has_name(Customer >> str): pass
    class has_email(Customer >> str): pass
    class has_loyalty_points(Customer >> int): pass
    class has_addresschangedbyemployer(Customer >> str): pass
    class has_addresschangedays(Customer >> int): pass
     
    # Add relationships (optional, if needed in ontology)
    class ordered_by(Order >> Customer): pass
    class contains_product(Order >> Product): pass

# Save the ontology to a file
output_path = "rule_ontology.owl"
ontology.save(file=output_path, format="rdfxml")
print(f"Ontology saved to {output_path}")
```

```python
from owlready2 import get_ontology
import json

# Specify the file path to your ontology
file_path = "C:\\Users\\rajes\\OneDrive\\Documents\\GitHub\\rajpersonal\\Graph\\rule_ontology.owl"
output_file = "C:\\Users\\rajes\\OneDrive\\Documents\\GitHub\\rajpersonal\\Graph\\rule_ontology.json"

try:
    # Load the ontology
    onto = get_ontology(file_path).load()
    print("Ontology loaded successfully!")

    # Prepare data for JSON export
    ontology_data = {
        "classes": []
    }

    # List all classes and their associated properties
    print("Classes and Attributes:")
    for cls in onto.classes():
        class_name = cls.iri.split("#")[-1] if "#" in cls.iri else cls.iri.split("/")[-1]  # Extract the class name
        class_data = {
            "name": class_name,
            "data_properties": [],
            "object_properties": []
        }

        # Add data properties associated with the class
        for prop in onto.data_properties():
            if cls in prop.domain:
                prop_name = prop.iri.split("#")[-1] if "#" in prop.iri else prop.iri.split("/")[-1]  # Extract the property name
                class_data["data_properties"].append(prop_name)

        # Add object properties associated with the class
        for prop in onto.object_properties():
            if cls in prop.domain:
                prop_name = prop.iri.split("#")[-1] if "#" in prop.iri else prop.iri.split("/")[-1]  # Extract the property name
                class_data["object_properties"].append(prop_name)

        print(f"Class: {class_name}")
        print(f"  Data Properties: {class_data['data_properties']}")
        print(f"  Object Properties: {class_data['object_properties']}")

        ontology_data["classes"].append(class_data)

    # Save to JSON file
    with open(output_file, "w", encoding="utf-8") as json_file:
        json.dump(ontology_data, json_file, indent=4)
    print(f"Ontology data saved to {output_file}")

except Exception as e:
    print(f"Error loading ontology: {e}")
```
